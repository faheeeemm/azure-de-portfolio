LIST COMPREHENSION

```python
orders = [1500, 800, 2200, 500, 3100, 950]

approved = [order for order in orders if order >= 1000]
rejected = [order for order in orders if order < 1000]
doubled = [order * 2 for order in orders]

print(approved)
print(rejected)
print(doubled)
```

```python
products = [
    {"name": "Shoes", "price": 1500},
    {"name": "Socks", "price": 200},
    {"name": "Jeans", "price": 2200},
    {"name": "Belt", "price": 400},
    {"name": "Jacket", "price": 3500},
]

expensive_products = [product["name"] for product in products if product["price"] > 1000]

print(expensive_products)
```

DICT COMPREHENSIONS

```python
products = [
    {"name": "Shoes", "price": 1500},
    {"name": "Socks", "price": 200},
    {"name": "Jeans", "price": 2200},
    {"name": "Belt", "price": 400},
    {"name": "Jacket", "price": 3500},
]

expensive_products = { product["name"]: product["price"] for product in products if product["price"] > 1000 }

print(expensive_products)
```

LAMBDA

```python
products = [
    {"name": "Shoes", "price": 1500},
    {"name": "Socks", "price": 200},
    {"name": "Jeans", "price": 2200},
    {"name": "Belt", "price": 400},
    {"name": "Jacket", "price": 3500},
]

sorted_prodcuts = sorted(products, key =lambda product: product["name"])
print(sorted_prodcuts)
```

```python
square = lambda x: x*x
print(square(5))
print(square(12))
```
