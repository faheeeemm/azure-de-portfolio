MAP/FILTER/REDUCE

```python
order_amounts = [250, 890, 45, 1200, 600, 30]

#exp_prod = lambda order : order>500
exp_prod = list(filter(lambda order : order>500, order_amounts))
print(exp_prod)

prod_amnt = list(map(lambda order: order/100, order_amounts))
print(prod_amnt)

final_orders = list(map(lambda order: order/100, filter(lambda order : order>500, order_amounts)))

print(final_orders)
```

```python
result = filter(lambda order: order > 500, order_amounts)
print(result)
```

```python
from functools import reduce 
order_amounts = [250, 890, 45, 1200, 600, 30]
total = reduce(lambda a, b: a + b, order_amounts,0)
print(total)
```

```python
def run_pipeline(pipeline_name):
    print(f"Starting: {pipeline_name}")
    try:
        result = 100/3
        print("Pipeline completed successfully")
    except ZeroDivisionError as e:
        print(f"Pipeline failed: {e}")
    finally:
        print(f"Closing connections for: {pipeline_name}")

run_pipeline("retail_etl")
```

Decorators

```python
from functools import wraps

def log_execution(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Starting: {func.__name__}")
        result = func(*args, **kwargs)
        print(f"Completed: {func.__name__}")
        return result
    return wrapper

@log_execution
def run_pipeline(pipeline_name):
    print(f"Processing: {pipeline_name}")

run_pipeline("retail_etl")
```

```python
#EXCERCISE
import time

def timer(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start =time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"Pipeline took {end-start} seconds")
        return result
    return wrapper

@log_execution
@timer 
def run_pipeline(pipeline_name):
    print(f"Processing: {pipeline_name}")

run_pipeline("retail_etl")
```

GENERATORS

```python
def generator_function():
    print("Starting")
    yield 1
    print("Middle")
    yield 2
    print("End")

gen = generator_function()
print(gen)
```

```python
# List comprehension - eager, builds full list now
squared_list = [x**2 for x in range(1000000)]

# Generator expression - lazy, one at a time
squared_gen = (x**2 for x in range(1000000))

print(squared_gen)
```
