LIST COMPREHENSION

```python
orders = [1500, 800, 2200, 500, 3100, 950]

approved = [order for order in orders if order >= 1000]
rejected = [order for order in orders if order < 1000]
doubled = [order * 2 for order in orders]

print(approved)
print(rejected)
print(doubled)
```

DICT COMPREHENSIONS

```python
products = [
    {"name": "Shoes", "price": 1500},
    {"name": "Socks", "price": 200},
    {"name": "Jeans", "price": 2200},
    {"name": "Belt", "price": 400},
    {"name": "Jacket", "price": 3500},
]

expensive_products = { product["name"]: product["price"] for product in products if product["price"] > 1000 }

print(expensive_products)
```

```python
def run_pipeline(pipeline_name):
    print(f"Starting: {pipeline_name}")
    try:
        result = 100/3
        print("Pipeline completed successfully")
    except ZeroDivisionError as e:
        print(f"Pipeline failed: {e}")
    finally:
        print(f"Closing connections for: {pipeline_name}")

run_pipeline("retail_etl")
```

Decorators

```python
from functools import wraps

def log_execution(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Starting: {func.__name__}")
        result = func(*args, **kwargs)
        print(f"Completed: {func.__name__}")
        return result
    return wrapper

@log_execution
def run_pipeline(pipeline_name):
    print(f"Processing: {pipeline_name}")

run_pipeline("retail_etl")
```

```python
#EXCERCISE
import time

def timer(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start =time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"Pipeline took {end-start} seconds")
        return result
    return wrapper

@log_execution
@timer 
def run_pipeline(pipeline_name):
    print(f"Processing: {pipeline_name}")

run_pipeline("retail_etl")
```
