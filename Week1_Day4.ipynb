```python
# CUSTOM CONTEXT MANAGER
class DBConnection:
    def __enter__(self):
        print("Connecting to DB!")
        return self
    
    def __exit__(self, exc_type, exc_value, traceback):
        print(f"exc_type: {exc_type}")
        print(f"exc_value: {exc_value}")
        print("Closing DB connection!")
        return True

with DBConnection() as db:
    print("Running Query!")
    raise ValueError("Query failed!")
print("After the with block")  # does this line ever run?
```

```python
with open("pipeline_logs.txt","w") as f:
    f.write("Job started\n")
    f.write("Job completed\n")
```

```python
def log_order(order_id, status, filename="pipeline_log.txt"):
    with open(filename,"a") as f:
        f.write(f"Order {order_id}: {status}\n")

log_order("OD001","Processed")
```

```python
import json

orders = [
    {"order_id": 101, "amount": 250, "status": "completed"},
    {"order_id": 102, "amount": 890, "status": "failed"}
]


with open("orders.json","w") as f:
    json.dump(orders,f)

with open("orders.json","r") as f:
    json_str = json.load(f)

json_str  = json.dumps(orders)
print(json_str)
print(type(json_str))

order = json.loads(json_str)
print(order)
print(type(order))
```

```python
import csv

with open("orders.csv","w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["order_id","amount","status"])
    writer.writerow([101,100,"placed"])
    writer.writerow([102,200,"shipped"])
    writer.writerow([103,300,"delivered"])
    writer.writerow([104,400,"placed"])
```

```python
orders = [
    {"order_id": 101, "amount": 250, "status": "completed"},
    {"order_id": 102, "amount": 890, "status": "failed"}

]

with open("orders2.csv","w", newline="") as f:
    fieldnames = ["order_id","amount","status"]
    writer = csv.DictWriter(f,fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(orders)
```

```python
with open("orders2.csv","r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        total = int(row['amount']) + 100
        print(row)
        print(total)
```
