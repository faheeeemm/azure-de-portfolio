```python
def calculate_order_value(quantity, unit_price):
    total = quantity * unit_price
    return total

result = calculate_order_value(5, 300)
print(result)

calculate_order_value(unit_price=300, quantity=5)
```

```python
def process_order(order_id, status="pending", priority="normal"):
    return {
        "order_id": order_id,
        "status": status,
        "priority": priority
    }

print(process_order(1001))
print(process_order(1002, status="approved"))
print(process_order(1003, status="approved", priority="high"))
```

```python
def log_pipeline_event(**details):
    print(type(details))
    for key, value in details.items():
        print(f"{key}: {value}")

log_pipeline_event(
    job_name="retail_etl",
    status="failed",
    error_count=3,
    region="north"
)
```

```python
def process_pipeline(pipeline_name,*tables,env="prod",**config):
    print(f"Pipeline: {pipeline_name}")
    print(f"Environment: {env}")
    print(f"Tables to process: {tables}")
    print(f"Additional config: {config}")

process_pipeline(
    "retail_etl",
    "orders", "products", "customers",
    env="dev",
    timeout=30,
    retry=3    
)
```

```python
class Product:
    def __init__( self,product_id, name, price):
        self.product_id=product_id
        self.name=name
        self.price=price

    def apply_discount(self, discount_percent):
        discounted_price = self.price - (self.price * discount_percent / 100)
        return discounted_price
    
    def __str__(self):
        return f"Product({self.product_id}, {self.name}, ₹{self.price})"

p1 = Product("P001", "Shoes", 1500)
p2 = Product("P002", "Shirt", 850)

print(p1.apply_discount(10))
print(p1)
```

```python
class Product:
    def __init__ (self, product_id, name, price):
        self.product_id=product_id
        self.name=name
        self.price=price
    
    def apply_discount(self, discount_percent):
        discounted_price=self.price - (self.price * discount_percent / 100)
        return discounted_price

class Clothing(Product):
    def __init__(self, product_id, name, price, size):
        super().__init__(product_id, name, price)
        self.size=size

c1 = Clothing("C001", "Jeans", 2000, "L")

print(c1.name)
print(c1.size)
print(c1.apply_discount(20))
    
```

```python
# EXCERCISE

class Electronics(Product):
    def __init__(self,product_id, name, price, warranty_years):
        super().__init__(product_id, name, price)
        self.warranty_years=warranty_years
    
    def warranty_info(self):
        return f"{self.name} comes with {self.warranty_years} year warranty"

e1 = Electronics("E001", "Laptop", 50000, 2)

print(e1.warranty_info())
```
