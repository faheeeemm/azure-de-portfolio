FUNCTIONS

```python
def calculate_order_value(quantity, unit_price):
    total = quantity * unit_price
    return total

result = calculate_order_value(5, 300)
print(result)

calculate_order_value(unit_price=300, quantity=5)
```

```python
def process_order(order_id, status="pending", priority="normal"):
    return {
        "order_id": order_id,
        "status": status,
        "priority": priority
    }

print(process_order(1001))
print(process_order(1002, status="approved"))
print(process_order(1003, status="approved", priority="high"))
```

```python
def calculate_total(*amounts):
    print(type(amounts))
    print(amounts)
    return sum(amounts)

print(calculate_total(1500, 2200, 800, 3100))
```

```python
def log_pipeline_event(**details):
    print(type(details))
    for key, value in details.items():
        print(f"{key}: {value}")

log_pipeline_event(
    job_name="retail_etl",
    status="failed",
    error_count=3,
    region="north"
)
```

```python
def process_pipeline(pipeline_name,*tables,env="prod",**config):
    print(f"Pipeline: {pipeline_name}")
    print(f"Environment: {env}")
    print(f"Tables to process: {tables}")
    print(f"Additional config: {config}")

process_pipeline(
    "retail_etl",
    "orders", "products", "customers",
    env="dev",
    timeout=30,
    retry=3    
)
```

```python
#EXERCISE

def validate_pipeline(pipeline_name,*tables,strict=False):
    if strict and len(tables)==0:
        print("Error: no tables defined")
    else:
        for table in tables:
            print(f"Validating table: {table}")
```

```python
validate_pipeline("retail_etl", "orders", "products", strict=True)
validate_pipeline("retail_etl", strict=True)
```
