DATATYPE

```python
# Retail order records — some have bad data
orders = [
    {"order_id": 1, "amount": "1500"},
    {"order_id": 2, "amount": "N/A"},
    {"order_id": 3, "amount": "2200"},
    {"order_id": 4, "amount": ""},
    {"order_id": 5, "amount": "3100"},
    {"order_id": 6, "amount": None}
]
```

```python
def process_orders(orders):
    for order in orders:
        try:
            order["amount"] = int(order["amount"])
            print(f"Order {order['order_id']} processed: {order['amount']}")
        except (ValueError, TypeError):
            print(f"Order {order['order_id']} skipped: bad data")
process_orders(orders)
```

CONTROL FLOW

```python
job_status = "pending"
error_count = 1

if job_status == "failed" and error_count >2:
    print("Critical: pipeline needs immediate attention")
elif job_status == "failed" and error_count <=2:
    print("Warning: pipeline failed with minor errors")
elif job_status == "success":
    print("Pipeline completed successfully")
else:
    print("Unknown status")
```

```python
attempts = 0
max_attempts = 3
job_status = "failed"

while job_status == "failed" and attempts < max_attempts:
    attempts += 1
    print(f"Attempt {attempts}: retrying pipeline job...")
    if attempts == 3:
        job_status = "success"

print(f"Final status: {job_status} after {attempts} attempts")
```
